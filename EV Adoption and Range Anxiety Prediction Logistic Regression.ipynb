```python
import pandas as pd
```

```python
df = pd.read_csv("EV_Adoption_and_Range_Anxiety_Dataset.csv")
```

```python
df.head()
```

```python
df["Will_Buy_EV"].value_counts(dropna=False)
```

```python
df.isna().sum()
```

```python
df["Annual_Income_USD"] = df["Annual_Income_USD"].fillna(df["Annual_Income_USD"].median())
```

```python
df["Daily_Commute_km"] = df["Daily_Commute_km"].fillna(df["Daily_Commute_km"].median())
```

```python
df["Environmental_Concern_Level"] = df["Environmental_Concern_Level"].fillna(df["Environmental_Concern_Level"].mean())
```

```python
df.isna().sum()
```

```python
df.duplicated().sum()
```

```python
df.info()
```

```python
df['Will_Buy_EV'] = df['Will_Buy_EV'].map({'Yes': 1, 'No': 0})
df.head()
```

```python
x = df.drop(["Buyer_ID","Will_Buy_EV"],axis=1)
y = df["Will_Buy_EV"]
```

```python
category_cols = [
    "Gender",
    "Daily_Commute_km",
    "Current_Car_Type",
    "Home_Charging_Possible",
    "Subsidy_Available",
    "Range_Anxiety_Level",
    "City_Type"
]
```

```python
x = pd.get_dummies(x,columns=category_cols,drop_first=True,dtype=int)
```

```python
x
```

```python
y
```

```python
from sklearn.model_selection import train_test_split
```

```python
x_train, x_test, y_train, y_test = train_test_split(
    x,y,test_size=0.20,stratify=y, random_state = 42
)
```

```python
from sklearn.preprocessing import StandardScaler
```

```python
scaler = StandardScaler()
```

```python
scaler
```

```python
x_train_scaled = scaler.fit_transform(x_train)
x_test_scaled = scaler.transform(x_test)
```

```python
from sklearn.linear_model import LogisticRegression
```

```python
model = LogisticRegression(
    max_iter = 1000,
    class_weight = "balanced"
)
```

```python
model
```

```python
model.fit(x_train_scaled,y_train)
```

```python
y_pred = model.predict(x_test_scaled)
```

```python
y_pred
```

```python
compariosn = pd.DataFrame({
    "Actual": y_test,
    "Prediction" : y_pred
})
```

```python
compariosn
```

```python
from sklearn.metrics import accuracy_score,precision_score,recall_score,f1_score
```

```python
accuracy = accuracy_score(y_test,y_pred)*100
precision = precision_score(y_test,y_pred)*100
recall =  recall_score(y_test,y_pred)*100
f1 = f1_score(y_test,y_pred)*100
```

```python
print(f"accuracy: {accuracy:.2f}%")
print(f"precision:{precision:.2f}%")
print(f"recall:{recall:.2f}%")
print(f"f1:{f1_score:.2f}%")
```

## Final Model Conclusion

The final **Logistic Regression model with `class_weight="balanced"`** was selected as the final model for predicting the target variable, where **0 = No** and **1 = Yes**.

The model achieved an overall **accuracy of 80.25%**. For the positive class (**Yes**), the model achieved **47% precision, 88% recall, and 61% F1-score**. The use of `class_weight="balanced"` significantly improved the detection of positive cases, increasing Class 1 recall from **65% to 88%** and F1-score from **53% to 61%**.

The model correctly identified **307 out of 350 actual Yes cases**, while 43 Yes cases were classified as No. Although the model produced some false positives, the improved recall indicates that it is effective at identifying positive cases.

Therefore, the **Logistic Regression model with balanced class weights** was selected as the final model because it provides a better balance between overall performance and detection of the minority **Yes** class.
